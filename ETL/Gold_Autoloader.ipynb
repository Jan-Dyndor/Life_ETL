```python
from pyspark.sql.functions import min, col, lit, max, day, month, year 
from config.core import config
```

```python
table_a = spark.table(f"{config.catalog.catalog_name}.{config.catalog.silver_schema}.{config.nbp.table_a}")
table_b = spark.table(f"{config.catalog.catalog_name}.{config.catalog.silver_schema}.{config.nbp.table_b}")
```

```python
gold_united = table_a.unionByName(table_b)
```

```python
gold_united = gold_united.withColumn("day", day("date"))\
    .withColumn("month", month("date"))\
    .withColumn("year", year("date"))
display(gold_united.head(5))
```

```python
gold_united.write.mode("overwrite").format("delta").saveAsTable(f"{config.catalog.catalog_name}.{config.catalog.gold_schema}.gold_table")
```
